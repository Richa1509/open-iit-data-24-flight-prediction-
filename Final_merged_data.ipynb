```python
import pandas as pd

# Load the data
sorted_combined = pd.read_csv('/content/sorted_combined_data.csv')  # Load sorted data
df = pd.read_csv('/content/final_preprocessed_data.csv')  # Load final preprocessed data

# Convert 'FlightDate' to datetime
df['date_converted'] = pd.to_datetime(df['FlightDate'], errors='coerce')

# Create an empty list to store the merged DataFrames for each month
merged_dfs = []

# Loop through the years and months
for year in [2022, 2023, 2024]:
    for month in range(1, 13):  # Months 1 to 12
        # Filter the DataFrame for the specific year and month
        month_df = df[(df['date_converted'].dt.year == year) & (df['date_converted'].dt.month == month)]

        # If there are no records for the current month, continue to the next month
        if month_df.empty:
            continue

        # Group by 'date_converted' and get the last entry
        month_df_last = month_df.groupby('date_converted', as_index=False).last()

        # Filter the sorted_combined data for the specific year and month
        month_weather = sorted_combined[(sorted_combined['date_converted'].dt.year == year) &
                                        (sorted_combined['date_converted'].dt.month == month)]

        # Ensure that the month_weather DataFrame is not empty
        if month_weather.empty:
            continue

        # Merge the DataFrames on 'date_converted'
        merged_month_df = pd.merge(month_df_last, month_weather, on='date_converted', how='inner')

        # Append the merged DataFrame to the list
        merged_dfs.append(merged_month_df)

# Combine all merged DataFrames into a single DataFrame
final_merged_df = pd.concat(merged_dfs, ignore_index=True)

# Now final_merged_df contains all the merged data for each month from Jan to Dec

```
